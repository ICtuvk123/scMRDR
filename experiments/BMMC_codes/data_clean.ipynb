```python
print("正在读取数据...")
atac = sc.read("./ATAC_counts_qc.h5ad")

# 2. 预处理 ATAC 数据
atac.var_names_make_unique() # 加上这句，防止源头报错
# ...这里可以放你原本的那些 chromosomes 过滤代码...
atac.layers['normalized'] = atac.X.copy()
atac.X = atac.layers['counts'].copy()

# 3. 生成 Gene Activity (最吃内存的一步)
print("正在计算 Gene Activity (内存峰值)...")
atac_gas = epi.tl.geneactivity(atac, "./gencode.v38.annotation.gtf", annotation='HAVANA')

# ==========================================
# 【关键操作】: 过河拆桥
# 生成了新数据 atac_gas 后，老数据 atac 就没用了
# 立即删除它，瞬间释放约 40GB 内存！
# ==========================================
print("正在释放旧数据内存...")
del atac
gc.collect() 
print("内存释放完毕！")
# ==========================================

# 4. 后续处理
# 使用官方推荐的去重方法，而不是删除
atac_gas.var_names_make_unique()

print("搞定！可以继续下一步了。")
```

```python
from scipy import sparse

# 1. 找到问题图层：layers['counts']
# 报错显示是这里出了问题
if 'counts' in atac_gas.layers:
    counts_layer = atac_gas.layers['counts']
    
    # 检查是不是 numpy.matrix
    if isinstance(counts_layer, np.matrix):
        print("检测到非法类型 numpy.matrix，正在修复...")
        # 转换为稀疏矩阵 (CSR)，既能解决报错，又能节省大量硬盘空间
        atac_gas.layers['counts'] = sparse.csr_matrix(counts_layer)
        print("修复完成！")
    else:
        print("类型正常，可能是其他原因，尝试强制转换...")
        # 强制转为稀疏矩阵保平安
        atac_gas.layers['counts'] = sparse.csr_matrix(counts_layer)

# 2. 再次尝试保存
# 【重要提醒】: 请务必保存到 /root/autodl-tmp/ 数据盘，
# 因为你之前的报错显示你还在尝试写入系统盘 ("./ATAC_gas.h5ad")，这会导致系统盘爆满！
save_path = "./ATAC_gas.h5ad"

print(f"正在保存到数据盘: {save_path} ...")
atac_gas.write(save_path)
print("保存成功！")
```

```python
import pandas as pd
df_hgnc = pd.read_csv("./group-471.csv",skiprows=1)
df_hgnc
```

```python
# 2. 清洗列名（保留这个好习惯，防止有空格）
df_hgnc.columns = df_hgnc.columns.str.strip()

# 再次检查列名（可选，看看还有没有问题）
print("清洗后的列名:", df_hgnc.columns.tolist())
cd_list = prot.var.index.tolist()
df_hgnc[['Previous symbols', 'Aliases']] = df_hgnc[['Previous symbols', 'Aliases']].fillna('')

def map_cd_name(cd):
    for _, row in df_hgnc.iterrows():
        approved = row['Approved symbol']
        prev = [s.strip() for s in str(row['Previous symbols']).split(',')]
        aliases = [s.strip() for s in str(row['Aliases']).split(',')]

        if cd == approved or cd in prev or cd in aliases:
            return approved
    return cd  # 若找不到，则保留原始名称

# 应用映射
mapped_cd_list = [map_cd_name(cd) for cd in cd_list]

# 结果放入DataFrame中
df_result = pd.DataFrame({'Original': cd_list, 'Mapped Gene Symbol': mapped_cd_list})

# 可选：查看哪些没有映射成功
unmapped = df_result[df_result['Original'] == df_result['Mapped Gene Symbol']]
```
